```python
tips["tip_ratio"] = tips["tip"] / tips["total_bill"]

# Create histogram bins manually
counts, bins = np.histogram(tips["tip_ratio"], bins=30)

# Colors: first 4 bars custom, rest gray
colors = ["blue", "green", "orange", "pink","purple","teal", "crimson","white"] + ["lightgray"] * (len(counts) - 4)

# Build bar centers for plotting
bin_centers = 0.5 * (bins[:-1] + bins[1:])

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=bin_centers,
        y=counts,
        marker=dict(
            color=colors,
            line=dict(color="black", width=3)   # black edges, thickness 3
        )
    )
)

fig.update_layout(
    width=800,
    height=500,
    title="Histogram with Custom Colors for First Four Bars",
    xaxis_title="tip_ratio",
    yaxis_title="count"
)

# Saves plot as HTML and opens it
fig.write_html("plot.html", auto_open=True)
```

```python
import yfinance as yf
import pandas as pd
import plotly.graph_objects as go

# ── 1. Download Live Data ──────────────────────────────────────────────────────
tickers = ['BAC', 'JPM', 'C', 'GS', 'MS']

start_date = '2006-01-01'
end_date = '2026-04-22'   # yfinance end is exclusive, so use next day for Apr 21, 2026

print("📥 Downloading live data from Yahoo Finance...")
raw = yf.download(
    tickers,
    start=start_date,
    end=end_date,
    auto_adjust=True,
    progress=False
)

df = raw['Close'].copy()
df.dropna(inplace=True)
df.index = pd.to_datetime(df.index)

print(f"✅ Data loaded: {df.shape[0]} trading days × {df.shape[1]} tickers")

# ── 2. Normalize so all series start at 1.0 ───────────────────────────────────
df_norm = df.div(df.iloc[0])

# ── 3. Weekly resample for smooth line look like your example ─────────────────
df_weekly = df_norm.resample('W').last().reset_index()
df_weekly.rename(columns={'Date': 'date'}, inplace=True)

# ── 4. Colors ──────────────────────────────────────────────────────────────────
colors = {
    'AMZN': '#2ca02c',
    'BAC':  '#d62728',
    'JPM':  '#1f77b4',
    'C':    '#9467bd',
    'GS':   '#17becf',
    'MS':   '#ff7f0e',
}


# ── 5. Build Figure ────────────────────────────────────────────────────────────
fig = go.Figure()

for ticker in tickers:
    fig.add_trace(go.Scatter(
        x=df_weekly['date'],
        y=df_weekly[ticker],
        mode='lines',
        name=ticker,
        line=dict(color=colors[ticker], width=1.8),
        hovertemplate=(
            f'<b>{ticker}</b><br>'
            'Date : %{x|%b %Y}<br>'
            'Value: %{y:.3f}<extra></extra>'
        )
    ))

# ── 6. Frames for Animation ────────────────────────────────────────────────────
dates = df_weekly['date'].tolist()
frames = []

for i in range(1, len(dates) + 1):
    frame_data = []
    for ticker in tickers:
        frame_data.append(go.Scatter(
            x=df_weekly['date'].iloc[:i],
            y=df_weekly[ticker].iloc[:i],
            mode='lines',
            name=ticker,
            line=dict(color=colors[ticker], width=1.8),
        ))
    frames.append(go.Frame(
        data=frame_data,
        name=str(dates[i - 1])[:10]
    ))

fig.frames = frames

# ── 7. Layout ──────────────────────────────────────────────────────────────────
fig.update_layout(
    # ── Title ──────────────────────────────────────────────────────────────────
    title=dict(
        text='📈 Normalized Stock Performance (Base = 1.0) — 2006 to 2026',
        x=0.5,
        xanchor='center',
        font=dict(size=18, color='#333333')
    ),

    # ── Axes ───────────────────────────────────────────────────────────────────
    xaxis=dict(
        title=dict(text='date', font=dict(size=13)),
        showgrid=True,
        gridcolor='white',
        gridwidth=1,
        tickformat='%b %Y',
        tickangle=0,
        tickfont=dict(size=11, color='#444444'),
        showline=False,
        zeroline=False,
    ),
    yaxis=dict(
        title=dict(text='Normalized Close (Start = 1.0)', font=dict(size=13)),
        showgrid=True,
        gridcolor='white',
        gridwidth=1,
        tickfont=dict(size=11, color='#444444'),
        showline=False,
        zeroline=False,
    ),

    # ── Background (match reference image: light blue-gray) ───────────────────
    plot_bgcolor='#dce6f5',
    paper_bgcolor='#dce6f5',

    # ── Legend ─────────────────────────────────────────────────────────────────
    legend=dict(
        title=dict(text='variable', font=dict(size=12)),
        x=1.01,
        y=1,
        xanchor='left',
        yanchor='top',
        bgcolor='rgba(220,230,245,0.0)',
        borderwidth=0,
        font=dict(size=12),
        traceorder='normal',
    ),

    # ── Size & Margins ─────────────────────────────────────────────────────────
    height=550,
    width=1400,
    margin=dict(l=60, r=120, t=80, b=120),
    hovermode='x unified',

    # ── Play / Pause / Reset Buttons ───────────────────────────────────────────
    updatemenus=[{
        'type': 'buttons',
        'showactive': True,
        'x': 0.5,
        'xanchor': 'center',
        'y': -0.22,
        'yanchor': 'bottom',
        'bgcolor': '#FFFFFF',
        'bordercolor': '#AAAAAA',
        'borderwidth': 1,
        'font': dict(size=13),
        'buttons': [
            {
                'label': '▶  Play',
                'method': 'animate',
                'args': [None, {
                    'frame': {'duration': 80, 'redraw': True},
                    'fromcurrent': True,
                    'transition': {'duration': 40, 'easing': 'linear'},
                    'mode': 'immediate',
                }]
            },
            {
                'label': '⏸  Pause',
                'method': 'animate',
                'args': [[None], {
                    'frame': {'duration': 0, 'redraw': False},
                    'mode': 'immediate',
                    'transition': {'duration': 0}
                }]
            },
            {
                'label': '⏮  Reset',
                'method': 'animate',
                'args': [[str(dates[0])[:10]], {
                    'frame': {'duration': 0, 'redraw': True},
                    'mode': 'immediate',
                    'transition': {'duration': 0}
                }]
            }
        ]
    }],

    # ── Slider ─────────────────────────────────────────────────────────────────
    sliders=[{
        'active': 0,
        'currentvalue': {
            'prefix': '📅 ',
            'visible': True,
            'xanchor': 'center',
            'font': dict(size=12, color='#333333')
        },
        'pad': {'t': 40, 'b': 5},
        'len': 0.88,
        'x': 0.06,
        'y': -0.12,
        'bgcolor': '#FFFFFF',
        'bordercolor': '#AAAAAA',
        'tickcolor': '#AAAAAA',
        'font': dict(size=9),
        'steps': [{
            'args': [[str(date)[:10]], {
                'frame': {'duration': 80, 'redraw': True},
                'mode': 'immediate',
                'transition': {'duration': 40}
            }],
            'label': pd.Timestamp(date).strftime('%b %Y'),
            'method': 'animate'
        } for date in dates]
    }]
)

# ── 8. Baseline Reference Line at y = 1.0 ─────────────────────────────────────
fig.add_hline(
    y=1.0,
    line_dash='dash',
    line_color='gray',
    line_width=1,
    opacity=0.6,
    annotation_text='Baseline (1.0)',
    annotation_position='bottom right',
    annotation_font=dict(size=10, color='gray')
)

# ── 9. Save & Show ─────────────────────────────────────────────────────────────
fig.write_html("normalized_stocks.html")
fig.write_html("plot.html", auto_open=True)
```
